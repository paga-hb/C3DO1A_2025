## The `kubectl` CLI Tool

In this notebook, we look at the very basics of using the `kubectl` CLI tool, which is used to communicate with a Kubernetes cluster.

- Make sure you have a Kubernetes cluster (Docker Desktop) running.
- Also make sure you have installed the `kubectl` tool on your computer.

## Set `kubectl`'s context to `docker-desktop`

- Since `kubectl` can be used to communicate with multiple Kubernetes cluster's, we need to tell `kubectl` which cluster we want to communicate with.

```python
!kubectl config get-contexts
!kubectl config use-context docker-desktop
!kubectl config current-context
```

```python
!kubectl version
```

## Show `kubectl` help

- `kubectl --help` shows what commands kubectl can run.
- `kubectl [command] --help` shows additional help for the command `[command]`.
  - e.g. `kubectl config --help` shows additional help for the command `config`.

```python
#!kubectl config --help
!kubectl --help
```
