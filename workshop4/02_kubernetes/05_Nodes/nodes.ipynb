## List Nodes

- The command `kubectl get nodes` is used to list the nodes running in a kubernetes cluster.
  - Without the `-o wide` option, provides basic information about the nodes.
  - With The `-o wide` option, provides additional information about the nodes.
- A Docker Desktop cluster always has 1 `control-plane` node (called `docker-desktop` here).
  - A Docker Desktop cluster doesn't any additional nodes, but usually a cluster would also have `worker` nodes.
- Here we see:
  - The `NAME` of each node.
  - The `STATUS` of each node (here `Ready` means the node is ready to use).
  - The `ROLES` of each node (`docker-desktop` is a Control Plane node).
  - The `AGE` of each node, i.e. how long the node has been available.
  - The `VERSION` or kubernetes each node is using.
  - The `INTERNAL-IP` addess of each node (the IP address each node has within the kubernetes cluster).
  - The `EXTERNAL-IP` addess of each node (the public IP address each node has outside the kubernetes cluster).
  - The `OS-IMAGE` used by each node (in this case Docker Desktop).
  - The `KERNEL-VERSION` of the OS used by each node.

## Get information about a node

- The command `kubectl get node [NodeName]` returns information about a node with name `[NodeName]`.
  - The `-o yaml` option returns the information as a YAML manifest (definition) for a node.

- Some important information contained in a node listing:
  ```bash
  apiVersion: v1
  kind: Node
  metadata:
    name: docker-desktop      <--- the name of the node
  status:
    addresses:
    - address: 192.168.65.3   <--- the node's IP addres
      type: InternalIP
    - address: docker-desktop <--- the node's hostname
      type: Hostname
    capacity:
      cpu: 8                           <--- number of CPU threads (number of CPU cores => divide by 2)
      ephemeral-storage: 1055762868Ki  <--- storage (disk size) in kibibytes.
      memory: 7917592Ki                <--- RAM in kibibytes
      pods: 110                        <--- maximum number of Pods that the ndoe can run
    images <list>                      <--- list of container images downloaded to the node
    nodeInfo:
      architecture: amd64
      containerRuntimeVersion: docker://27.4.0           <--- container runtime version
      kernelVersion: 5.15.167.4-microsoft-standard-WSL2
      kubeProxyVersion: v1.30.5                          <--- kube proxy version
      kubeletVersion: v1.30.5                            <--- kubelet version
      operatingSystem: linux                             <--- operating system type
      osImage: Docker Desktop
  ```

````python
#!kubectl get node docker-desktop
!kubectl get node docker-desktop -o yaml
````

## Describe a node

- The command `kubectl describe nodes` returns detailed information about all nodes.
- The command `kubectl describe node [NodeName]` returns detailed information about a node with name `[NodeName]`.
- A node description contains the same, but more detailed, information as a node's YAML listing above.
- In addition, a node description contains a list of important `Events` that have taken place (at the bottom of the listing).
- In this case we have `Events: <none>`, but in a production cluster we might have events as below ... 
 
  ```bash
  Events:
    Type    Reason                   Age                From             Message
    ----    ------                   ----               ----             -------
    Normal  Starting                 50m                kube-proxy       
    Normal  Starting                 51m                kubelet          Starting kubelet.
  ```

````python
#!kubectl describe nodes
!kubectl describe node docker-desktop
````
