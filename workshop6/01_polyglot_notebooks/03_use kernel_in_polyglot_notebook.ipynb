# Using a Custom Kernel in a Ployglot Notebook

- In this notebook, we use *connect to* our custom Python kernel from a Polyglot Notebook.
  - The Python kernel was registered as a Jupyter kernel in a previous notebook.
- This allows us to use Python on a Polyglot Notebook cell. 

---

## Connect to Jupyter Python Kernel

- **Make sure you have:**
  - Installed the [dotnet interactive](https://github.com/dotnet/interactive) tool.
  - Installed the [Polyglot Notebooks Extension](https://marketplace.visualstudio.com/items?itemName=ms-dotnettools.dotnet-interactive-vscode).
  - Registered a Python Environment as a Jupyter Python Kernel.
  - Selected `.NET Interactive` as the Jupyter Notebook Kernel in the top right of this notebook.
- Let's connect to the Jupyter Python Kernel we registered in the previous notebook.
  - Remember, we registred the Python Environment as `.conda`, which we supply as a parameter to `--kernel-spec` below.
  - The name `python` we supply as a parameter to `--kernel-name` below, is the name that will be displayed when we choose to use the environment in a notebook cell.
  - Note that we have chosen `csharp - C# Script` for our `Code` cell below (i.e. we are running `C#` code in the cell).
  - The *special* comment below is specific to Polyglot Notebooks, and is used to connect to our Python Environment in this case.
    - **When connecting to our Python Kernel, we have to use a `C#` `Code` cell as below.**   

```python
#!connect jupyter --kernel-name python --kernel-spec .conda
```

---

## Verify the Python Kernel is Working

- Verify the Python Kernel is working by executing the cell below (which should print `Hello World from Python!`).
- Notice that we have chosen the `python` in our `Code` cell below (hence the notebook cell is running `Python` code).
  - The name `python` comes from the `--kernel-name` we used in the cell above.
  - It's using the Python Environment we registered under `.conda` (via the `--kernel-spec` settings above).
    - **When using our `Python` kernel in a notebook `Code` cell, we make sure we have selected it as in the cell below.**

```python
print(f"Hello World from Python!")
```

---

## Running other Languages in a Polyglot Notebook

- The major benefit of a Polyglot Notebook, is of course that we can run multiple programming languages (or *kernels*) in the same Notebook.
  - Although, we can only run one language (not multiple languages) in a cell.
- Above we selected `python` in a `Code` cell, so we could run `Python` code (just as in a *normal* Notebook).
- Below we have selected `csharp` in a `Code` cell, so we can run `C#` code.
  - Notice we can change the language by clicking on the chosen language in the bottom right of a code cell.
  - We can also toggle the cell between `Code` and `markdown` by clicking on the `Code/markdown` selection in the bottom right of a cell.

```python
Console.WriteLine("Hello World from C#!");
```
