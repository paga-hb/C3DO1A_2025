## Install the `dotnet interactive` Tool

Make sure that the `dotnet tools` folder is on your `PATH` (environment variable).

- On Windows (note it's probably better to set this manually on Windows):

    ```bash
    setx Path "%PATH%;%USERPROFILE%\.dotnet\tools
    ```

- On Linux/Mac:

    ```bash
    echo 'export PATH="$HOME/.dotnet/tools:$PATH"' >> ~/.bashrc  # or ~/.zshrc
    source ~/.bashrc  # or source ~/.zshrc
    ```


The install the [dotnet interactive](https://github.com/dotnet/interactive) tool by executing the cell below.

````python
!dotnet tool install --global Microsoft.dotnet-interactive
# !dotnet tool list -g
!dotnet interactive jupyter install
````
