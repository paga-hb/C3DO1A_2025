# Register a Python Environment as a Jupyter Kernel

- In this notebook, we register our current Python Environment as a Jupyter Kernel.
- This allows us to:
  - Refer to this specific Python Environment in a Jupyter Notebook.
  - Connect to this specific Python Environment from a Polyglot notebook (which enables us to run Python cells in a Polyglot Notebook).

---

## Register our Current Python Environment as Jupyter Kernel

Registering a Python Environment for Jupyter Notebook:
- Let's register our current Python Environment (in the `.conda` folder) as a Jupyter Kernel.
- This let's us connect to the  Python Environment from our Polyglot Notebooks so that we can run Python cells within the notebook.

The code below:
- First removes a Jupyter kernel (*kernelspec*) named `.conda` from the list of registred Jupyter kernels.
- Then a Jupyter kernel is registered under the name `.conda` for the current Python Environment (`.conda` shown in the top right of this notebook).
- The actual specification for the kernel (*kernelspec*) is stored in a file called `kernel.json` found in the folder `.conda` in the listing below.
- So now we can refer to the registred name `.conda` to reference the Python Environment (also called `.conda`) when *connecting* to it from a Polyglot Notebook. 

```python
#!conda env list
!jupyter kernelspec remove .conda -y
!python -m ipykernel install --user --name=.conda --display-name "Python (.conda)"
!jupyter kernelspec list
```
