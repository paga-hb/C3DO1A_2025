## Start a SQL Server Container

The [SQL Server Image](https://hub.docker.com/r/microsoft/mssql-server) is fairly easy to use when creating a container.
- We use the command `docker run -d --rm mcr.microsoft.com/mssql/server:2022-latest` to:
  - Create and start a container (`docker run`),
  - Instantiate the container based on the SQL Server Image (`mcr.microsoft.com/mssql/server:2022-latest`).
  - Run the container in detached mode (`-d`).
  - Automatically delete the container when it is stopped (`--rm`).
- We also:
  - Set `--name` and `--hostname` to e.g. `sqlserver` (this name can be used as the DNS name in connection strings).
  - Map host port `1433` to container port `1433` using the `-p` flag.
  - Set two environment variables using the `-e` flag:
    - `ACCEPT_EULA=Y` so that we don't have to manually confirm the license agreement.
    - `MSSQL_SA_PASSWORD=P@55w0rd` to set the password for the `sa` user.
  - Map a named volume `splserver_data` (on the host) to container folder `/var/opt/mssql` so that our databases are preserved when we delete the container.

We can use the following connection string to connect to e.g. the `master` database on the SQL Server instance running inside the container.
  - Note that we are using `localhost` since we are connecting from the host computer (not from within the container).

```bash
Data Source=localhost,1433;
Initial Catalog=master;
User Id=sa;
Password=P@55w0rd;
Encrypt=Yes;
TrustServerCertificate=Yes;
MultipleActiveResultSets=True;
```

````python
!docker run -d --rm --name sqlserver --hostname sqlserver -p 1433:1433 -e ACCEPT_EULA=Y -e MSSQL_SA_PASSWORD=P@55w0rd -v sqlserver_data:/var/opt/mssql mcr.microsoft.com/mssql/server:2022-latest
!docker ps
````

## Create a Database with a Table

- We are using a connection string to connect to the `TestDatabase`:

    ```bash
    Data Source=localhost,1433;Initial Catalog=TestDatabase;User Id=sa;Password=P@55w0rd;Encrypt=Yes;TrustServerCertificate=Yes;MultipleActiveResultSets=True;
    ```
- We are creating one entity class `User`:

    ```csharp
    public class User
    {
        public int Id { get; set; }
        public string Name { get; set; }
    }
    ```
- We create a `DbContext` called `MyDbContext`, define a `DbSet<User>` property called `Users`, and use SQLServer with our connection string:

    ```csharp
    public class MyDbContext : DbContext
    {
        public DbSet<User> Users { get; set; }

        protected override void OnConfiguring(DbContextOptionsBuilder optionsBuilder)
        {
            // Connection string to SQL Server (running in the Container)
            optionsBuilder.UseSqlServer(CONNECTION_STRING);
        }
    }
    ```

- Finally, we create an instance of our `MyDbContext` and use it to create the database, followed by listing all users (which is an empty list):

    ```csharp
    // Method for listing all users in the Users table
    private void ListUsers(MyDbContext context)
    {
        // List all rows in the 'Users' table
        Console.WriteLine($"\nUsers\n------------\nId\tName\n------------");
        var users = context.Users.ToList();
        if(users.Count == 0)
            Console.WriteLine($"<empty>");
        else
            foreach(var user in users) { Console.WriteLine($"{user.Id}\t{user.Name}"); }
    }

    // Create an instance of our 'MyDbContext' which will connect to the DBMS
    // Notice it's being created in a "using" scope so that it will disconnect from the DBMS when it goes out of scope
    using (var context = new MyDbContext())
    {
        // Create the database and table if they don't exist
        context.Database.EnsureCreated();
        Console.WriteLine($"Created database 'TestDatabase' with table 'Users'.");

        // List all rows in the 'Users' table
        ListUsers(context);
    }
    ```

````python
using Microsoft.EntityFrameworkCore;
using System;
using System.Linq;

// Define a connection string to connect to the SQL Server Container
static var CONNECTION_STRING="Data Source=localhost,1433;Initial Catalog=TestDatabase;User Id=sa;Password=P@55w0rd;Encrypt=Yes;TrustServerCertificate=Yes;MultipleActiveResultSets=True;";

// Define a "User" entity class that maps to the "Users" table
public class User
{
    public int Id { get; set; }
    public string Name { get; set; }
}

// Define the DbContext that manages the database connection and table mapping
public class MyDbContext : DbContext
{
    public DbSet<User> Users { get; set; }

    protected override void OnConfiguring(DbContextOptionsBuilder optionsBuilder)
    {
        // Connection string to SQL Server (running in the Container)
        optionsBuilder.UseSqlServer(CONNECTION_STRING);
    }
}

// Method for listing all users in the Users table
private void ListUsers(MyDbContext context)
{
    // List all rows in the 'Users' table
    Console.WriteLine($"\nUsers\n----------------\nId\tName\n----------------");
    var users = context.Users.ToList();
    if(users.Count == 0)
        Console.WriteLine($"<empty>");
    else
        foreach(var user in users) { Console.WriteLine($"{user.Id}\t{user.Name}"); }
}

// Create an instance of our 'MyDbContext' which will connect to the DBMS
// Notice it's being created in a "using" scope so that it will disconnect from the DBMS when it goes out of scope
using (var context = new MyDbContext())
{
    // Create the database and table if they don't exist
    context.Database.EnsureCreated();
    Console.WriteLine($"Created database 'TestDatabase' with table 'Users'.");

    // List all rows in the 'Users' table
    ListUsers(context);
}
````

## Insert a Row

- We create an instance of our `MyDbContext` and use it to insert a `User` into the `Users`table, followed by listing all users (which contains one user):

    ```csharp
    User insertedUser = null;
    using (var context = new MyDbContext())
    {
        // Insert a new user into the Users table
        var user = new User { Name = "John Doe" };
        context.Users.Add(user);
        context.SaveChanges();
        Console.WriteLine("Inserted 'John Doe' into the 'Users' table.");

        // Query the table to verify the inserted row
        insertedUser = context.Users.FirstOrDefault(u => u.Name == "John Doe");
        if (insertedUser != null)
        {
            Console.WriteLine($"Found User: ID = {insertedUser.Id}, Name = {insertedUser.Name}");
        }
        context.Entry(insertedUser).State = EntityState.Detached;

        // List all rows in the 'Users' table
        ListUsers(context);
    }
    ```

````python
User insertedUser = null;
using (var context = new MyDbContext())
{
    // Insert a new user into the Users table
    var user = new User { Name = "John Doe" };
    context.Users.Add(user);
    context.SaveChanges();
    Console.WriteLine("Inserted 'John Doe' into the 'Users' table.");

    // Query the table to verify the inserted row
    insertedUser = context.Users.FirstOrDefault(u => u.Name == "John Doe");
    if (insertedUser != null)
    {
        Console.WriteLine($"Found User: ID = {insertedUser.Id}, Name = {insertedUser.Name}");
    }
    context.Entry(insertedUser).State = EntityState.Detached;

    // List all rows in the 'Users' table
    ListUsers(context);
}
````
