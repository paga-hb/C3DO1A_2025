## Configure Azure Blob Storage Settings

We'll use the **default account** for the Azure Storage Emulator below:
- STORAGE_ACCOUNT_NAME: `devstoreaccount1`
- STORAGE_ACCESS_KEY: `Eby8vdM02xNOcqFlqUwJPLlmEtlCDXJ1OUzFT50uSRZ6IFsuFq2UVErCz4I6tq/K1SZFPTOtr/KBHBeksoGMGw==`

The default URLs to the three different storage types provided by the Azure Storage Emulator are:
- BLOB_URL: http://127.0.0.1:10000/devstoreaccount1
- QUEUE_URL: http://127.0.0.1:10001/devstoreaccount1
- TABLE_URL: http://127.0.0.1:10002/devstoreaccount1

Below, we are also configuring a connection string that can be used to access the default Storage Account:

- CONNECTION_STRING:

    ```bash 
    AccountName=devstoreaccount1;
    AccountKey=Eby8vdM02xNOcqFlqUwJPLlmEtlCDXJ1OUzFT50uSRZ6IFsuFq2UVErCz4I6tq/K1SZFPTOtr/KBHBeksoGMGw==;
    DefaultEndpointsProtocol=http;
    BlobEndpoint=http://127.0.0.1:10000/devstoreaccount1;
    QueueEndpoint=http://127.0.0.1:10001/devstoreaccount1;
    TableEndpoint=http://127.0.0.1:10002/devstoreaccount1;
    ```

Let's run the cell below to set this up.

````python
var STORAGE_ACCOUNT_NAME = "devstoreaccount1";
var STORAGE_ACCESS_KEY = @"Eby8vdM02xNOcqFlqUwJPLlmEtlCDXJ1OUzFT50uSRZ6IFsuFq2UVErCz4I6tq/K1SZFPTOtr/KBHBeksoGMGw==";
var STORAGE_CONTAINER_NAME = "videos";
var SCHEME = "http";
var DOMAIN = "127.0.0.1";
int BLOB_PORT = 10000;
int QUEUE_PORT = 10001;
int TABLE_PORT = 10002;

var BLOB_URL = $"{SCHEME}://{DOMAIN}:{BLOB_PORT}/{STORAGE_ACCOUNT_NAME}";
var QUEUE_URL = $"{SCHEME}://{DOMAIN}:{QUEUE_PORT}/{STORAGE_ACCOUNT_NAME}";
var TABLE_URL = $"{SCHEME}://{DOMAIN}:{TABLE_PORT}/{STORAGE_ACCOUNT_NAME}";

var CONNECTION_STRING =
    $"AccountName={STORAGE_ACCOUNT_NAME};" +
    $"AccountKey={STORAGE_ACCESS_KEY};" +
    $"DefaultEndpointsProtocol={SCHEME};" +
    $"BlobEndpoint={BLOB_URL};" +
    $"QueueEndpoint={QUEUE_URL};" +
    $"TableEndpoint={TABLE_URL};";

STORAGE_ACCOUNT_NAME = Environment.GetEnvironmentVariable("AZURE_STORAGE_ACCOUNT_NAME") ?? STORAGE_ACCOUNT_NAME;
STORAGE_ACCESS_KEY = Environment.GetEnvironmentVariable("AZURE_STORAGE_KEY") ?? STORAGE_ACCESS_KEY;
STORAGE_CONTAINER_NAME = Environment.GetEnvironmentVariable("AZURE_STORAGE_CONTAINER_NAME") ?? STORAGE_CONTAINER_NAME;
CONNECTION_STRING = Environment.GetEnvironmentVariable("AZURE_STORAGE_CONNECTION_STRING") ?? CONNECTION_STRING;

Console.WriteLine($"STORAGE_ACCOUNT_NAME: {STORAGE_ACCOUNT_NAME}");
Console.WriteLine($"STORAGE_ACCESS_KEY: {STORAGE_ACCESS_KEY}");
Console.WriteLine($"STORAGE_CONTAINER_NAME: {STORAGE_CONTAINER_NAME}");
Console.WriteLine();
Console.WriteLine($"CONNECTION_STRING: {CONNECTION_STRING}");
Console.WriteLine();
Console.WriteLine($"BLOB_URL: {BLOB_URL}");
Console.WriteLine($"QUEUE_URL: {QUEUE_URL}");
Console.WriteLine($"TABLE_URL: {TABLE_URL}");
````

## Connecting to Azure BLOB Storage

**Note! Make sure you have started the Blob Emulator.**

````python
// Create a BlobServiceClient instance
var sharedKeyCredential = new StorageSharedKeyCredential(STORAGE_ACCOUNT_NAME, STORAGE_ACCESS_KEY);
var blobServiceClient = new BlobServiceClient(new Uri(BLOB_URL), sharedKeyCredential); // Connect using URL and key
// var blobServiceClient = new BlobServiceClient(CONNECTION_STRING); // Connect using connection string
// Console.WriteLine(blobServiceClient.GetProperties()); // Test connection
````

---

## Listing BLOB Containers and BLOBs

````python
// Method for listing containers with their blobs
private void ListContainersAndBlobs()
{
    var containers = blobServiceClient.GetBlobContainers();
    if(containers.Count() == 0)
    {
        Console.WriteLine("<empty>");
        return;
    }
    foreach (var container in containers)
    {
        Console.WriteLine($"- {container.Name}");
        var containerClient = blobServiceClient.GetBlobContainerClient(container.Name);
        var blobs = containerClient.GetBlobs();
        foreach (var blob in blobs)
        {
            Console.WriteLine($"   - {blob.Name}");
        }
    }
}

Console.WriteLine("\nCurrent Container List with Blobs:");
ListContainersAndBlobs();
````

---

## Creating a BLOB Container

````python
// # Create container
var blobContainerClient = blobServiceClient.GetBlobContainerClient(STORAGE_CONTAINER_NAME);
if (blobContainerClient.Exists())
{
    Console.WriteLine($"\nContainer '{STORAGE_CONTAINER_NAME}' already exists.");
}
else
{
    Console.WriteLine($"\nAdding container '{STORAGE_CONTAINER_NAME}'.");
    // blobServiceClient.CreateBlobContainer(STORAGE_CONTAINER_NAME);
    blobContainerClient.CreateIfNotExists();
}

Console.WriteLine("\nCurrent Container List with Blobs:");
ListContainersAndBlobs();
````

---

## Uploading/Updating an Object to a BLOB Container

````python
// # Upload/Update file as blob
var localFilePath = "fixtures/SampleVideo_1280x720_1mb.mp4";
var blobName = "SampleVideo_1.mp4";
Console.WriteLine($"\nUploading File '{localFilePath}' to Azure Blob Storage container '{STORAGE_CONTAINER_NAME}' as '{blobName}'");
blobContainerClient.CreateIfNotExists();
var blobClient = blobContainerClient.GetBlobClient(blobName);
using (var fileStream = File.OpenRead(localFilePath))
{
    blobClient.Upload(fileStream, overwrite: true);  // Overwrite if the file exists
}
Console.WriteLine($"✅ File '{localFilePath}' uploaded to Azure Blob Storage container '{STORAGE_CONTAINER_NAME}' as '{blobName}'");

Console.WriteLine("\nCurrent Container List with Blobs:");
ListContainersAndBlobs();
````

---

## Downloading an Object from a BLOB Container

````python
// # Download blob as file
var localFilePath = "./SampleVideo_1.mp4";
var blobName = "SampleVideo_1.mp4";
Console.WriteLine($"\nDownloading '{blobName}' from Azure Blob Storage container '{STORAGE_CONTAINER_NAME}' to File '{localFilePath}'");
blobContainerClient.CreateIfNotExists();
var blobClient = blobContainerClient.GetBlobClient(blobName);
using (var fileStream = File.OpenWrite(localFilePath))
{
    blobClient.DownloadTo(fileStream);
}
Console.WriteLine($"✅ Blob '{blobName}' downloaded from Azure Blob Storage container '{STORAGE_CONTAINER_NAME}' to File '{localFilePath}'");

Console.WriteLine("\nCurrent Container List with Blobs:");
ListContainersAndBlobs();
````

---

## Deleting an Object from a BLOB Container

````python
// # Delete blob
Console.WriteLine($"\nDeleting blob '{blobName}' from Azure Blob Storage container '{STORAGE_CONTAINER_NAME}'");
blobClient.Delete();

Console.WriteLine("\nCurrent Container List with Blobs:");
ListContainersAndBlobs();
````

---

## Deleting a BLOB Container

````python
// # Delete container
Console.WriteLine($"\nDeleting container '{STORAGE_CONTAINER_NAME}'");
blobServiceClient.DeleteBlobContainer(STORAGE_CONTAINER_NAME);

Console.WriteLine("\nCurrent Container List with Blobs:");
ListContainersAndBlobs();
````
