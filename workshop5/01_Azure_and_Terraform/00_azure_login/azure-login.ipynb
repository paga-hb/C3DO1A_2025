## Azure CLI command output format

- You can configure the output format for each Azure CLI command using the `--output [Format]` option, where `[Format]` is the desired format.
- You can also use the abbreviated version `-o [Format]` instead of `--output [Format]`.
- The default output format is `json`, which will be used if no output format is specified for a command:
- Allowed values for `[Format]` are:
  -  `json`       output is returned as a JSON string
  -  `jsonc`      output is returned as a JSON string with color-coded values
  -  `none`       no output (except for errors and warnings)
  -  `table`      output is returned as an ASCII table of values
  -  `tsv`        output is returned as tab-separated values
  -  `yaml`       output is returned as a YAML document
  -  `yamlc`      output is returned as a YAMLC document with color-coded values
- The default output format is `json`, which will be used if no output format is explicitly specified for a command.

````python
#!az version
#!az version -o json
#!az version -o jsonc
#!az version -o none
!az version -o table
#!az version -o tsv
#!az version -o yaml
#!az version -o yamlc
````

## Viewing Your Azure Subscriptions

- To view all your Azure subscriptions, you can use the command `az account list --output table`
  - The `Name` column contains the name of your subscription (of which one entry should be `Azure for Students`).
  - The `CloudName` column contains the cloud's name (which will ne be `AzureCloud`).
  - The `SubscriptionId` column contains the unique ID of your subscription.
  - The `TenantId` column contains your unique tentant ID.
  - The `State` column shows the state of your subscription (which should be `Enabled`).
  - The `IsDefault` column shows which of your subscriptions is the default subscription.
    - This should be `True` for your `Azure for Students` subscription.
    - Otherwise, you can set it to `True` (see below).

Run the cell below to show your Azure subscriptions.

````python
!az account list --output table
````

## Setting Your Default Azure Subscription

- To set your default Azure subscription, you can use the command `az account set --subscription "<SubscriptionId>"`
  - The `<SubscriptionId>` is the unique Subscription ID for the subscription you want to set at the default subscription.
    - You want to set this to the Subscription ID of your `Azure for Students` subscription.

**Change `<SubscriptionId>` in the cell below to your `SubscriptionId` for `Azure for Students` from the cell above.**

````python
!az account set --subscription "<SubscriptionId>"
````

## Istalling VSCode Extensions for Azure

- There are numerous VSCode Extensions available for VSCode.
  - Go to the Extensions view (Windows/Linux: `Ctrl + Shift + X`, Mac: `Cmd + Shift + X`).
  - Search for `azure` in the search box.
- Some Azure Extensions you might try (currently, some of the Extensions are in *Preview*):
  - [Azure Tools](https://marketplace.visualstudio.com/items?itemName=ms-vscode.vscode-node-azure-pack): An **Extension Pack** (i.e. collection of Extensions) with a good collection of basic Azure Extensions.
    - Also see the [Getting Started](https://code.visualstudio.com/docs/azure/gettingstarted) guide.
  - [Azure Resources](https://marketplace.visualstudio.com/items?itemName=ms-azuretools.vscode-azureresourcegroups): Extension for working with Azure Resources (including Azure Resource Groups).
  - [Azure Kubernetes Service](https://marketplace.visualstudio.com/items?itemName=ms-kubernetes-tools.vscode-aks-tools): Extension for working with Azure Kubernetes Service (AKS).
    - Also see the [Introduction](https://azure.github.io/vscode-aks-tools/index.html) and [Installation](https://azure.github.io/vscode-aks-tools/installation.html#installation) instructions.
  - [Azurite](https://marketplace.visualstudio.com/items?itemName=Azurite.azurite): An emulator for emulating Azure Storage accounts (Blob, Queue and Table) for local development purposes.
    - Also see the [Azurite Documentation](https://learn.microsoft.com/en-us/azure/storage/common/storage-use-azurite) and how to [Install Azurite](https://learn.microsoft.com/en-us/azure/storage/common/storage-use-azurite?tabs=visual-studio%2Cblob-storage#install-azurite).
    - **Note!**
      - I would recommend installing [NodeJS](https://nodejs.org), [NPM](https://www.npmjs.com) and finally the NPM Azurite package (`npm install -g azurite`), which means you'll be able to use Azurite from your terminal, otherwise you'll only be able to work with Azurite via the VSCode Extension or by using the VSCode Command Palette (Windows/Linux: `Ctrl + Shift + P`, Mac: `Cmd + Shift + P`).
      - When doing so, from the [Download Page](https://nodejs.org/en/download), I would recommend installing NodeJS and NPM via the [NVM](https://github.com/nvm-sh/nvm) or [FNM](https://github.com/Schniz/fnm) Version Managers, which allows you to easily switch between different versions of NodeJS and NPM.
