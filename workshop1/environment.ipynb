## Docker Desktop

Let's verify Docker Desktop is set up correctly.

- Search for `Docker Desktop` among the installed applications on your computer and start it.
  - If you are on Linux, and have chosen to install `Docker Engine` instead of `Docker Desktop`, ensure the  `Docker Engine` daemon is running.
- In the bottom left of `Docker Desktop's` user interface, wait until the status text `Engine starting` changes to `Engine running`.
  - If you are on Windows, and prompted to set up Windows Subsystem for Linux (WSL), follow the instructions to do so.
- Click the `cog wheel` icon in the top right of `Docker Desktop's` user interface.
  - Under the `General` tab, make sure the checkbox is checked for `Start Docker Desktop when you sign in to your computer`.
  - Under the `Kubernetes` tab, make sure `Enable Kubernetes` is enabled.
  - Click the `Apply & restart` button in the lower right of `Docker Desktop's` user interface.
  - When `Docker Desktop` has restarted, wait until the status text `Engine starting` changes to `Engine running` again.
  - Also, to the right of this text, in the bottom of `Docker Desktop's` user interface, wait until you see the text `Kubernetes running`.

With the setting above, `Docker Desktop` should start when your computer boots, and run automatically run `Docker Engine` and `Kubernetes`.

- **Note!**: If you didn't check the checkbox for `Start Docker Desktop when you sign in to your computer`, you need to start `Docker Desktop` manually after your computer has booted, before you can use `Docker`, `Docker Compose` or the local `Kubernetes` cluster.

## Docker Desktop Kubernetes

**Note that Docker Desktop must be running to use Docker Desktop's Kubenetes Cluster**

**Note that Kubectl must be installed on your computer to run the commands below**

Let's make sure we can interact with Docker Desktop's Kubernetes cluster from our terminal.

- Open Visual Studio Code's built-in terminal (Windows/Linux: `Ctrl + J`, Mac: `Cmd + J`).
- Run the commands below in the terminal.
  - `kubectl config use-context docker-desktop`
    - which should return `Switched to context "docker-desktop"` 
  - `kubectl config current-context`
    - which should return `docker-desktop`
  - `kubectl get nodes`
    - which should return a row containing the text `docker-desktop`.

You can also run the cell below to verify the commands can be run via your operating system's shell.

```python
!kubectl config use-context docker-desktop
!kubectl config current-context
!kubectl get nodes
```
